# Primary & foreign Key Validation 

```python
tables = {
    "customers": customers,
    "products" : products,
    "orders" : orders,
    "order_item" : order_items,
    "payments" : payments,
    "shipping" : shipping,
    "reviews" : reviews
}

pk_columns = {
    "customers": "customer_id",
    "products" : "product_id",
    "orders" : "order_id",
    "order_item": "order_item_id",
    "payments" : "payment_id",
    "shipping" : "shipping_id",
    "reviews" : "review_id"
}

# check for nulls in primary key 

for table_name, df in tables.items():
    pk = pk_columns[table_name]
    print(f"{table_name} → NULL PKs:", df[pk].isna().sum())
```

```python
# check for duplicate in primary keys

for table_name, df in tables.items():
    pk = pk_columns[table_name]
    print(f"{table_name} → Duplicate PKs:", df[pk].duplicated().sum())
```

foreign key Validation

```python
# 1. Customers → Locations
print("Customers → Locations")
print(customers["location_id"].isin(locations["location_id"]).value_counts())
```

```python
# 2. Orders → Customers
print("Orders → Customers")
print(orders["customer_id"].isin(customers["customer_id"]).value_counts())
```

```python
# 3. Order Items → Orders
print("Order Items → Orders")
print(order_items["order_id"].isin(orders["order_id"]).value_counts())
```

```python
# 4. Order Items → Products
print("Order Items → Products")
print(order_items["product_id"].isin(products["product_id"]).value_counts())
```

```python
# 5. Payments → Orders
print("Payments → Orders")
print(payments["order_id"].isin(orders["order_id"]).value_counts())
```

```python
# 6. Shipping → Orders
print("Shipping → Orders")
print(shipping["order_id"].isin(orders["order_id"]).value_counts())
```

```python

# 7. Shipping → Locations
print("Shipping → Locations")
print(shipping["shipping_location_id"].isin(locations["location_id"]).value_counts())
```

```python

# 9. Reviews → Order Items
print("Reviews → Order Items")
print(reviews["order_item_id"].isin(order_items["order_item_id"]).value_counts())
```

```python
# 10. Reviews → Customers
print("Reviews → Customers")
print(reviews["customer_id"].isin(customers["customer_id"]).value_counts())
```

# Final check 

```python
# FINAL NULL CHECK

for table_name, df in tables.items():
    print(f"{table_name}: {df.isna().sum().sum()} NULL values")
```

```python
# FINAL DUPLICATE CHECK

for table_name, df in tables.items():
    print(f"{table_name}: {df.duplicated().sum()} duplicate rows")
```

```python
for table_name, df in tables.items():
    print(f"{table_name}: {df.shape[0]} rows × {df.shape[1]} columns")
```
